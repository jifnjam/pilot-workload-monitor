# Phase 1: Understanding the data (N-back test EEG data)

### Set up data path and libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = "../data/raw/workload_dataset/data_n_back_test/eeg/eeg.parquet"
```

### Load in data

```python
df = pd.read_parquet(DATA_PATH)
df.head()
```

### Checking that the data for the subjects, test conditions, and phases are all there.

```python
print("Shape: ", df.shape)
print("Columns: ", list(df.columns))
print("Subjects: ", df['subject'].unique())
print("Test conditions: ", df['test'].unique())
print("Phases: ", df['phase'].unique())
```

```python
df.info()
```

### Reorganizing the dataset

```python
counts = df.groupby(['subject', 'test', 'phase']).size().reset_index(name="n_rows")
print(f"Total subject/test/phase combinations: {len(counts)}")
print(f"Expected if complete: {16 * 3 *3} (16 subjects, 3 tests, 3 phases)")
counts.shape
```

#### Picking a random subject, test, and phase

```python
sample = df[(df["subject"] == "subject_01") & (df["test"] == 1) & (df["phase"] == 2)]
print(sample)
```

### Plotting the dataset via matplotlib

```python
possible_channel_cols = ["EEG.Counter", "EEG.Interpolated", "EEG.AF3", "EEG.F7", "EEG.F3",
                         "EEG.FC5", "EEG.T7", "EEG.P7", "EEG.O1", "EEG.O2", "EEG.P8", "EEG.T8",
                         "EEG.FC6", "EEG.F4", "EEG.F8", "EEG.AF4", "EEG.RawCq", "EEG.Battery"][2:3]

print("Plotting columns: ", possible_channel_cols)

fig, ax = plt.subplots(figsize=(10,4))
sample[possible_channel_cols].reset_index(drop=True).plot(ax=ax)
ax.set_title("Sample EEG segment - subject_01, test=2 (med), phase=2 (task)")
ax.set_xlabel("Sample index")
ax.set_ylabel("Amplitude")
plt.tight_layout()
plt.show()
```
